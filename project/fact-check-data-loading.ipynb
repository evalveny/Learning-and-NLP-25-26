# Fake news data loading ([PubHealth dataset](https://huggingface.co/datasets/bigbio/pubhealth))

> Predefined train, test and validation sets may be used.

The dataset we will be using mainly contains the label of some online news, which can be "false", "true", "mixture" or "unknown".

There will be 3 different text entries: "claim", "explanation" and "main_text".

- `claim`: this column contains some quote regarding any topic, which can be true, false, a mixture of both or with unkown veracity.
- `main_text`: this column contains the information of some news that generally act as source for fact-checking the claim.
- `explanation`: this column contains the explanations made by journalists to confirm or deny the claim.

> You will have already defined the `train`, `test` and `validation` sets. Feel free to use any of both folders (pairs or source) for your interest.

```python
import pandas as pd

# load the data paths
pair_path = "./data/pubhealth_bigbio_pairs/"  # this df will only contain ["claim_id", "claim", "explanation", "label"]
source_path = "./data/pubhealth_source/"      # this df will also contain "main_text"

columns = ["claim_id", "claim", "explanation", "main_text", "label"] # columns with most information

# mapping dictionary
map_labels = {0: "true", 1: "false", 2: "unproven", 3: "mixture"}

# read parquet files
source_train_df = pd.read_parquet(source_path + "train/train.parquet")[columns]
source_val_df = pd.read_parquet(source_path + "val/val.parquet")[columns]
source_test_df = pd.read_parquet(source_path + "test/test.parquet")[columns]

# map class id to strings
source_train_df['label_class'] = source_train_df['label'].map(map_labels)
source_val_df['label_class'] =   source_val_df['label'].map(map_labels)
source_test_df['label_class'] =  source_test_df['label'].map(map_labels)

# display df
source_train_df.head()
```
